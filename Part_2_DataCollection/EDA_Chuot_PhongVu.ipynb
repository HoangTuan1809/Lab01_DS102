# Tổng quan về bộ dữ liệu
- Nguồn dữ liệu được thu thập từ API danh mục chuột máy tính của Phong Vũ: https://discovery.tekoapis.com/api/v2/search-skus-v2
- Kích thước: 500 điểm dữ liệu (sản phẩm) với 8 trường thông tin
- Các thuộc tính của bộ dữ liệu:
   + ProductId: Mã định danh trên hệ thống API
   + sku: Mã định danh sản phẩm trong kho hàng
   + Name: Tên sản phẩm
   + Brand: Thương hiệu
   + Retail_price: Giá niêm yết (VND)
   + Latest_price: Giá bán thực tế sau giảm giá (VND)
   + Discount_Percent: Phần trăm giảm giá
   + Discount_Amount: Số tiền giảm giá (VND)

```python
import pandas as pd

df = pd.read_csv('Chuot_PhongVu.csv')
print(f"Tổng điểm dữ liệu: {len(df)}")
display(df.head())
```

```python
print("--- THÔNG TIN CÁC CỘT DỮ LIỆU ---")
df.info()

print("\n--- KIỂM TRA GIÁ TRỊ NULL ---")
print(df.isnull().sum())

print("\n--- KIỂM TRA TRÙNG LẶP ---")
print(f"Số dòng bị trùng lặp: {df.duplicated().sum()}")
```

```python
cols = ['Retail_Price', 'Latest_Price', 'Discount_Percent', 'Discount_Amount']

print("--- THỐNG KÊ MÔ TẢ ---")
display(df[cols].describe().T[['mean', '50%', 'min', 'max', 'std']].rename(columns={'50%': 'median'}))
```

```python
print("--- TOP 10 THƯƠNG HIỆU PHỔ BIẾN NHẤT ---")
brand_summary = df['Brand'].value_counts().reset_index()
brand_summary.columns = ['Thương hiệu', 'Số lượng sản phẩm']
brand_summary['Tỷ lệ (%)'] = (brand_summary['Số lượng sản phẩm'] / len(df) * 100).round(2)

display(brand_summary.head(10))
```
